# TVP-VAR specification comparison

Notebook này so sánh baseline `λβ=0.999` với robustness `λβ=0.997`. Các tham số `κσ=0.96`, `p=1` và `H=10` được giữ cố định.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

from tvp_connectedness import (
    compare_specifications,
    load_volatility_data,
    run_specification,
)
```

```python
data = load_volatility_data(ROOT / 'data/model2_conditional_volatility.csv')

baseline = run_specification(
    data, name='baseline_0999', lambda_beta=0.999,
    kappa_sigma=0.96, initial_covariance=0.01, horizon=10,
)
robustness = run_specification(
    data, name='robustness_0997', lambda_beta=0.997,
    kappa_sigma=0.96, initial_covariance=0.01, horizon=10,
)
```

```python
comparison = compare_specifications([baseline, robustness])
comparison.round(4)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 9), sharex=True)

axes[0].plot(baseline.connectedness.tci, label='λβ=0.999')
axes[0].plot(robustness.connectedness.tci, label='λβ=0.997', alpha=0.8)
axes[0].set_ylabel('TCI (%)')
axes[0].set_title('Dynamic Total Connectedness')
axes[0].legend()
axes[0].grid(alpha=0.3)

for name in data.columns:
    axes[1].plot(baseline.connectedness.net[name], label=f'{name}, λ=.999')
    axes[1].plot(robustness.connectedness.net[name], linestyle='--', alpha=0.7, label=f'{name}, λ=.997')
axes[1].axhline(0, color='black', linewidth=1)
axes[1].set_ylabel('NET (%)')
axes[1].set_title('Dynamic Net Connectedness')
axes[1].legend(ncol=2)
axes[1].grid(alpha=0.3)
plt.tight_layout()
```

## Kết luận

Cấu trúc transmitter–receiver trung bình bền vững giữa hai specifications: BTC và SJC là net transmitters, VN-Index là net receiver. Vai trò cuối mẫu nhạy hơn với forgetting factor, đặc biệt đối với VN-Index.
