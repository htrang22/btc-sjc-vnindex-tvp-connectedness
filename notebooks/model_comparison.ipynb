# TVP-VAR robustness and sensitivity analysis

Chạy từ đầu và so sánh `λβ`, `κσ` và forecast horizon. Baseline: `λβ=0.999`, `κσ=0.96`, `p=1`, `H=10`, `P₀=0.01`.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from tvp_connectedness import compare_specifications, load_volatility_data, run_specification
data = load_volatility_data(ROOT / 'data/model2_conditional_volatility.csv')
print(data.shape, data.index.min(), data.index.max())
```

```python
settings = [
    dict(name='baseline_0999', lambda_beta=0.999, kappa_sigma=0.96, horizon=10),
    dict(name='robustness_0997', lambda_beta=0.997, kappa_sigma=0.96, horizon=10),
    dict(name='kappa_094', lambda_beta=0.999, kappa_sigma=0.94, horizon=10),
    dict(name='kappa_098', lambda_beta=0.999, kappa_sigma=0.98, horizon=10),
    dict(name='horizon_05', lambda_beta=0.999, kappa_sigma=0.96, horizon=5),
    dict(name='horizon_20', lambda_beta=0.999, kappa_sigma=0.96, horizon=20),
]
results = {s['name']: run_specification(data, initial_covariance=0.01, **s) for s in settings}
comparison = compare_specifications(list(results.values()))
comparison.round(4)
```

## Forgetting-factor sensitivity

```python
comparison.loc[['baseline_0999', 'robustness_0997', 'kappa_094', 'kappa_098']].round(4)
```

## Horizon sensitivity: H = 5, 10, 20

```python
comparison.loc[['horizon_05', 'baseline_0999', 'horizon_20']].round(4)
```

## Local-stability diagnostics

```python
baseline = results['baseline_0999']
unstable = baseline.stability.loc[~baseline.stability['stable']]
unstable_tci = baseline.connectedness.tci.loc[unstable.index]
stable_tci = baseline.connectedness.tci.loc[baseline.stability['stable']]
print('Unstable observations:', len(unstable))
print('Unstable episodes:', len(baseline.episodes))
print('Mean TCI on unstable dates:', round(unstable_tci.mean(), 4))
print('Mean TCI on stable dates:', round(stable_tci.mean(), 4))
display(baseline.episodes)
display(unstable.join(baseline.connectedness.tci).join(baseline.connectedness.net.add_prefix('NET_')))
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 12), sharex=True)
for key in ['horizon_05', 'baseline_0999', 'horizon_20']:
    r = results[key]; axes[0].plot(r.connectedness.tci, label=f'H={r.horizon}')
axes[0].set_ylabel('TCI (%)'); axes[0].set_title('Horizon sensitivity'); axes[0].legend()
for key in ['kappa_094', 'baseline_0999', 'kappa_098']:
    r = results[key]; axes[1].plot(r.connectedness.tci, label=f'κσ={r.model.kappa_sigma}')
axes[1].set_ylabel('TCI (%)'); axes[1].set_title('Covariance forgetting-factor sensitivity'); axes[1].legend()
axes[2].plot(baseline.stability['spectral_radius'], label='Spectral radius')
axes[2].axhline(1, color='black', linestyle='--', label='Boundary')
axes[2].scatter(unstable.index, unstable['spectral_radius'], color='red', s=20, label='Local instability')
axes[2].set_ylabel('Spectral radius'); axes[2].set_title('Baseline local stability'); axes[2].legend()
for axis in axes: axis.grid(alpha=0.3)
plt.tight_layout()
```

## Kết luận
